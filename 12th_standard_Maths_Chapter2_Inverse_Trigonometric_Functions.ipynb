```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_arcsin = np.linspace(-1 , 1, 500)
x_arccos = np.linspace(-1, 1, 500)
x_arctan = np.linspace(-10, 10, 500)

```

```python
fig, axs = plt.subplots(1,3, figsize=(18,5))


axs[0].plot(x_arcsin, np.arcsin(x_arcsin), label='arcsin(x)', color='blue')
axs[0].set_title('Inverse Sine Function')
axs[0].grid(True)
axs[0].legend()



axs[1].plot(x_arccos, np.arccos(x_arccos), label='arccos(x)', color='green')
axs[1].set_title('Inverse Cosine Function')
axs[1].grid(True)
axs[1].legend()

axs[2].plot(x_arctan, np.arctan(x_arctan), label='arctan(x)', color='red')
axs[2].set_title('Inverse Tangent Function')
axs[2].grid(True)
axs[2].legend()
```

**Note**

**arcsin(x) is same as asin(x)**
**arccos(x) is same as acos(x)**
**arctan(x) is same as atan(x)**




Lets try to compare a function and its invers


```python
x = np.linspace(-np.pi/2, np.pi/2, 500)
y = np.sin(x)


plt.plot(x, y, label='sin(x)')
plt.plot(y, x, label='asin(x)', linestyle='--')
plt.title('sin(x) vs asin(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
```

```python
from ipywidgets import interact, Dropdown
import warnings

warnings.filterwarnings('ignore')


```

```python
def plot_inverse_trig(func_name):
    plt.figure(figsize=(8, 5))

    if func_name == 'arcsin':
        x = np.linspace(-1, 1, 500)
        y = np.arcsin(x)
        plt.plot(x, y, label='arcsin(x)', color='blue')
        plt.title('Inverse Sine Function: arcsin(x)')
        plt.xlabel('x')
        plt.ylabel('y')
        plt.grid(True)

    elif func_name == 'arccos':
        x = np.linspace(-1, 1, 500)
        y = np.arccos(x)
        plt.plot(x, y, label='arccos(x)', color='green')
        plt.title('Inverse Cosine Function: arccos(x)')
        plt.xlabel('x')
        plt.ylabel('y')
        plt.grid(True)

    elif func_name == 'arctan':
        x = np.linspace(-10, 10, 500)
        y = np.arctan(x)
        plt.plot(x, y, label='arctan(x)', color='red')
        plt.title('Inverse Tangent Function: arctan(x)')
        plt.xlabel('x')
        plt.ylabel('y')
        plt.grid(True)

    plt.legend()
    plt.show()
```

```python
interact(plot_inverse_trig, func_name=Dropdown(
    options=['arcsin', 'arccos', 'arctan'],
    description='Function:',
    value='arcsin'
));
```
