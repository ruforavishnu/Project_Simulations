```python
import numpy as np
```

```python
A = np.array([[2,3],
              [1,4]])

B = np.array([[1,2,3],
              [0, -1, 4],
              [5, 2, 0]])
```

```python
print('Matrix A:\n', A )
```

```python
print('Determinant of A:\n', np.linalg.det(A))
```

```python
print('Matrix B:\n', B )
```

```python
print('Determinant of B:\n', np.linalg.det(B))
```

**Area of Triangle Using Determinant**

```python
def area_of_triangle(p1, p2, p3):
  mat = np.array([[p1[0], p1[1], 1],
                  [p2[0], p2[1], 1],
                  [p3[0], p3[1], 1],
                  ])

  return 0.5 * abs(np.linalg.det(mat))
```

```python
p1, p2, p3 = (0,0), (4,0), (2,3)
```

```python
print('Area of triangle: ', area_of_triangle(p1, p2, p3))
```

**Collinearity of Points**

```python
def are_collinear(p1, p2, p3):
  mat = np.array([
                  [p1[0], p1[1], 1],
                  [p2[0], p2[1], 1],
                  [p3[0], p3[1], 1],
                  ])

  return np.linalg.det(mat) == 0

```

```python
print('Collinear? ', are_collinear((0,0) , (1,1), (2,2) )  )
```

```python
print('Collinear? ', are_collinear((0,0) , (4,0), (2,3) )  )
```

**Volume of a Parallelepiped**

```python
a = np.array([1, 0, 0])
b = np.array([0, 2, 0])
c = np.array([0, 0, 3])


mat = np.column_stack([a, b, c])
volume = abs(np.linalg.det(mat))

```

```python
print('Volume of a parellelepipid: ', volume)
```

**Plot the Triangle (Area from Determinant)**

```python
import matplotlib.pyplot as plt
```

```python
def plot_triangle(p1, p2, p3):
  x = [p1[0], p2[0], p3[0], p1[0]]
  y = [p1[1], p2[1], p3[1], p1[1]]

  plt.figure(figsize=(5,5))
  plt.fill(x, y, color='lightblue', alpha=0.5, label='Triangle')
  plt.plot(x, y, 'o-', color='blue')


  for i, p in enumerate([p1, p2, p3], start=1):
    plt.text(p[0]+0.1, p[1]+0.1, f"P{i}{p}")


  plt.title(f"Triangle with Area = {area_of_triangle(p1, p2, p3):.2f}")
  plt.xlabel('X-axis')
  plt.ylabel('Y-axis')
  plt.grid(True)
  plt.axis('equal')
  plt.legend()
  plt.show()


```

```python
plot_triangle( (0,0), (4,0), (2,3))
```
