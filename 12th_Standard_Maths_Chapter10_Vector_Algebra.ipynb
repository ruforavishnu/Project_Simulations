**Problem:**

Suppose we are pushing a heavy sofa across a room.

Instead of pushing straight, the person sometimes pushes at an angle.

We want to understand:
- How much of the force actually moves the sofa forward?
- How much effort is wasted?
- Which angle is the most efficient?

We will use vector algebra to analyze this.

```python
import numpy as np
import matplotlib.pyplot as plt

```

```python
#force
F = 100 #newton


angles_deg = np.linspace(0,90, 100)
angles_rad = np.radians(angles_deg)


forward_force = F * np.cos(angles_rad)
sideways_force = F * np.sin(angles_rad)


```

```python
plt.figure()


plt.plot(angles_deg, forward_force)
plt.xlabel('Angles (degrees)')
plt.ylabel('Forward Force (N)')
plt.title('Forward force vs Angle')


plt.grid()
plt.show()
```

**now plot wasted effort**

```python
plt.figure()

plt.plot(angles_deg, sideways_force)
plt.xlabel('Angle (degrees)')
plt.ylabel('Wasted force (N)')
plt.title('Wasted force vs Angle')
plt.grid()
plt.show()

```

**now lets plot it both**

```python
plt.figure()

plt.plot(angles_deg, forward_force,  label='Forward (Useful)  force')
plt.plot(angles_deg, sideways_force,  label='Wasted force')

plt.xlabel('Angle (degrees)  ')
plt.ylabel('Force (N)')

plt.title('Useful Force vs Wasted Force')
plt.legend()
plt.grid()
plt.show()
```

```python
theta = np.radians(45)


Fx = F * np.cos(theta)
Fy = F * np.sin(theta)


plt.figure()

#org force vector
plt.quiver(0,0,    Fx, Fy,      angles='xy',       scale_units='xy',       scale=1)

#forward component
plt.quiver(0,0,    Fx, 0,      angles='xy',       scale_units='xy',       scale=1)


#side component
plt.quiver(Fx,0,    0, Fy,      angles='xy',       scale_units='xy',       scale=1)



plt.xlim(0,   F)
plt.ylim(0,   F)



plt.xlabel('Forward direction')
plt.ylabel('Sideways direction')
plt.title('Force decomposition at 45°')

plt.grid()
plt.show()


```

**now lets try efficiency**

```python
efficiency = (forward_force / F) * 100





plt.figure()

plt.plot(angles_deg, efficiency)

plt.xlabel("Angle (degrees)")
plt.ylabel("Efficiency (%)")
plt.title("Efficiency vs Angle")

plt.grid()
plt.show()
```
