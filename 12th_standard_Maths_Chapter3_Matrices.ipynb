```python
import numpy as np
```

```python
A = np.array([[2,3],
             [1,4]])


B = np.array([[5,2],
             [0,7]])

```

```python
print('Matrix A: \n', A)
print('\n\nMatrix B: \n', B)
```

Basic Matrix Operations

```python
print('A + B: \n', A + B)
```

```python
print('A - B: \n', A - B)
```

```python
print('A *3 (scalar multiplication): \n', 3* A)
```

```python
print('A *3 (scalar multiplication): \n', A*3)
```

```python
print('A dot B (matrix multiplication): \n', np.dot(A,B))
```

Transpose, Determinant, Inverse

```python
print('Transpose of A: \n', A.T)
```

```python
det_A = np.linalg.det(A)
print('Determinant of A: ', det_A)
```

```python
if det_A != 0:
  inv_A = np.linalg.inv(A)
  print('Inverse of A: \n', inv_A)
else:
  print('A is not invertible')
```

Solve a System of Equations

2x + 3y = 8


x + 4y = 7

```python
coeff = np.array([[2,3],
                  [1,4]])

rhs = np.array([8,7])

solution = np.linalg.solve(coeff, rhs)
print('Solution for [x,y]: ', solution)

```
