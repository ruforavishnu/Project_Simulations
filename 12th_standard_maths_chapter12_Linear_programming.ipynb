**Problem:**

Comparison of Cost Required for Muscle-Building Protein Intake During a Workout Routine Over 30 Days: Chicken vs Cherupayar.

We need to find out how much money it will cost in a month




```python
import numpy as np
import matplotlib.pyplot as plt


```

```python
protein_required = 120



#chicken
chicken_protein_per_100g = 31
chicken_cost_per_100g = 27


#cherupayar
cherupayar_protein_per_100g = 7
cherupayar_cost_per_100g = 8


```

```python
chicken_needed = (protein_required / chicken_protein_per_100g) * 100

cherupayar_needed = (protein_required / cherupayar_protein_per_100g) * 100




chicken_daily_cost = (chicken_needed/100)  * chicken_cost_per_100g
cherupayar_daily_cost = (cherupayar_needed/100)  * cherupayar_cost_per_100g






print("Chicken needed per day: ", round(chicken_needed,2)   , "g")
print("Chicken daily cost: Rs", round(chicken_daily_cost))


print("Cherupayar needed per day: ", round(cherupayar_needed,2)   , "g")
print("Cherupayar daily cost: Rs", round(cherupayar_daily_cost))



```

```python
days = np.arange(1, 31)


chicken_cost_30_days = chicken_daily_cost * days
cherupayar_cost_30_days = cherupayar_daily_cost * days


```

```python
chicken_total_cost = chicken_cost_30_days[-1]
cherupayar_total_cost = cherupayar_cost_30_days[-1]

chicken_total_protein = protein_required * 30
cherupayar_total_protein = protein_required * 30
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.plot(days,  chicken_cost_30_days)
plt.title('Chicken diet cost over 30 days')
plt.xlabel('Days')
plt.ylabel('Total Cost (₹)')
plt.grid()

plt.text(5, chicken_total_cost*0.5,
         f"Total Cost: ₹{round(chicken_total_cost,2)}\n"
         f"Protein: {chicken_total_protein}g",
         bbox=dict())

plt.subplot(1,2,2)
plt.plot(days, cherupayar_cost_30_days)
plt.title('Cherupayar cost over 30 days')
plt.xlabel('Days')
plt.ylabel('Total cost (₹)')
plt.grid()


plt.text(5, cherupayar_total_protein*0.5,
         f"Total Cost: ₹{round(cheru_total_cost,2)}\n"
         f"Protein: {cherupayar_total_protein}g",
         bbox=dict())

plt.tight_layout()
plt.show()

```
