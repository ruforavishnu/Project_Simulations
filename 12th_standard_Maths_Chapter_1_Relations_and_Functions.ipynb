```python
import matplotlib.pyplot as plt
import networkx as nx
```

```python
A = {1, 2, 3}
B = {4, 5, 6}



R = {(1,4), (2,5), (3,6)}

```

```python
def draw_relation(A, B, R):
  G = nx.DiGraph()

  for a in A:
    G.add_node(f"A{a}", bipartite=0)

  for b in B:
    G.add_node(f"B{b}", bipartite=1)

  for a,b in R:
    G.add_edge(f"A{a}", f"B{b}")


  pos = {}

  pos.update((f"A{a}", (0, -a)) for a in A)
  pos.update((f"B{b}", (1, -b)) for b in B)

  nx.draw(G, pos, with_labels=True, arrows=True, node_size=1500, node_color='lightblue', font_size=12)
  plt.title('Graph of Relation R')
  plt.show()


```

```python
draw_relation(A,B, R)
```

```python
def is_reflexive(A, R):
  return all( (a,a) in R for a in A)

def is_symmetric(R):
  return all( (b,a) in R for (a,b) in R)

def is_transitive(R):
  return all( (a,c) in R for (a, b1) in R for (b2,c) in R if b1 == b2)

```

```python
print("Reflexive: ", is_reflexive(A, R))
print('Symmetric: ', is_symmetric(R))
print('Transitive: ', is_transitive(R))
```

```python
def is_function(A, R):
  domain_values = [a for (a,b) in R]
  return len(set(domain_values)) == len(A)


def is_one_one(R):
  mapped_values = [b for (a,b) in R]
  return len(set(mapped_values)) == len(mapped_values)


def is_onto(A, B, R):
  return set(b for (a,b) in R) == B

```

```python
print('Is a function: ', is_function(A,R))
print('Is one-one (injective): ', is_one_one(R))
print('Is onto (surjective): ', is_onto(A, B, R))
```
