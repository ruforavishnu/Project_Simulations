```python
import random
import matplotlib.pyplot as plt
from collections import Counter


outcomes =['HH','HT','TH','TT']
experiments = [''.join(random.choices(['H','T'], k=2 )) for _ in range(10_000)]

counts = Counter(experiments)
total = sum(counts.values())


empirical_probs = {k: v/total  for k,v in counts.items()}

theoretical_probs = {'HH' : 0.25, 'HT': 0.25, 'TH':0.25, 'TT':0.25}

```

```python
labels = outcomes
empirical_values = [empirical_probs.get(k,0) for k in labels]
theoretical_values = [theoretical_probs[k] for k in labels]

```

```python
x = range(len(labels))

plt.figure(figsize=(8,5))
plt.bar(x, empirical_values, width=0.4, label='Empirical', align='center')
plt.bar([i + 0.4 for i in x], theoretical_values, width=0.4, label='Theoretical', align='center')
plt.xticks([i + 0.2 for i in x], labels)
plt.title('Emperical vs Theoretical Probabilities (2 Coin Tosses)')
plt.xlabel('Outcome')
plt.ylabel('Probability')
plt.legend()
plt.grid(True)
plt.show()
```

 3 Coin Tosses

```python
# Simulate 10,000 tosses of 3 coins
experiments = [''.join(random.choices(['H', 'T'], k=3)) for _ in range(10000)]
```

```python
from itertools import product
outcomes = [''.join(p) for p in product('HT', repeat=3)]  # ['HHH', 'HHT', ..., 'TTT']
```

```python

counts = Counter(experiments)
total = sum(counts.values())


empirical_probs = {k: v/total  for k,v in counts.items()}


```

```python
# Since all outcomes are equally likely, each has probability 1/8
theoretical_probs = {outcome: 1/8 for outcome in outcomes}
```

```python
labels = outcomes
empirical_values = [empirical_probs.get(k,0) for k in labels]
theoretical_values = [theoretical_probs[k] for k in labels]
```

```python
x = range(len(labels))

plt.figure(figsize=(8,5))
plt.bar(x, empirical_values, width=0.4, label='Empirical', align='center')
plt.bar([i + 0.4 for i in x], theoretical_values, width=0.4, label='Theoretical', align='center')
plt.xticks([i + 0.2 for i in x], labels)
plt.title('Emperical vs Theoretical Probabilities (3 Coin Tosses)')
plt.xlabel('Outcome')
plt.ylabel('Probability')
plt.legend()
plt.grid(True)
plt.show()
```

Simulating Card Draws

```python
import random
import matplotlib.pyplot as plt
from collections import Counter

```

```python
suits = ['Hearts','Diamonds','Clubs','Spades']
ranks = ['2', '3', '4','5','6','7','8','9','10','Jack','Queen','King','Ace']

deck = [rank + ' of ' + suit for suit in suits for rank in ranks]
print(deck)
```

```python
draws = [random.choice(deck) for _ in range(10_000)]

```

```python
print(draws)
```

```python
def is_ace(card):
  return 'Ace' in card

def is_red(card):
  return 'Hearts' in card or 'Diamonds' in card

def is_face(card):
  return any(face in card for face in ['Jack','Queen','King'])

def is_spade(card):
  return 'Spades' in card

```

```python
ace_count = sum(is_ace(card) for card in draws)
red_count = sum(is_red(card) for card in draws)
face_count = sum(is_face(card) for card in draws)
spade_count = sum(is_spade(card) for card in draws)

```

```python
print('ace_count:', ace_count)
print('red_count:', red_count)
print('face_count:', face_count)
print('spade_count:', spade_count)


```

```python
total = len(draws)
print(total)
```

```python
probabilities = {

                 'Ace':ace_count / total,
                 'Red card': red_count / total,
                 'Face card': face_count/total,
                 'Spade' : spade_count/total
}
```

```python
theoretical = {
    'Ace' : 4/52,
    'Red card': 26/52,
    'Face card': 12/52,
    'Spade' : 13/52
}
```

```python
labels = list(probabilities.keys())
empirical_vals = [probabilities[k] for k in labels]
theoretical_vals = [theoretical[k] for k in labels]

```

```python
x = range(len(labels))

plt.figure(figsize=(8,5))
plt.bar(x, empirical_vals, width=0.4, label='Empirical', align='center')
plt.bar([i + 0.4 for i in x], theoretical_vals, width=0.4, label='Theoretical', align='center')
plt.xticks([ i + 0.2 for i in x], labels)
plt.title('Card draw simulation : Empirical vs Theoretical Probability')
plt.ylabel('Probability')
plt.legend()
plt.grid(True)
plt.show()
```
