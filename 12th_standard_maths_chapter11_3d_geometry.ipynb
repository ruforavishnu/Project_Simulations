A person is standing inside a room at a certain point in space.

One wall of the room is represented as a plane.

We want to find:
- The shortest distance from the person to the wall
- The exact point on the wall closest to the person

We will also visualize this in 3D.

```python
#The shortest distance from a point to a plane is along the perpendicular.

#This means we need to drop a perpendicular from the point to the plane.


#we can use ax + by + cz + d = 0

```

```python
import numpy as np

import matplotlib.pyplot as plt


a, b , c, d = 2,3,4, -20

P = np.array( [4,  2,  3])


```

```python
numerator = abs(a * P[0]     + b*P[1]  +   c*P[2]   +    d)
denominator = np.sqrt(   a**2       +     b**2      + c**2)



distance = numerator / denominator


print("Shortest distance to wall:  ", distance)

```

$$
\lambda = \frac{a x_0 + b y_0 + c z_0 + d}{a^2 + b^2 + c^2}
$$

so

$$
Q = (x_0, y_0, z_0) - \frac{a x_0 + b y_0 + c z_0 + d}{a^2 + b^2 + c^2} (a, b, c)
$$

```python
lam = (a *P[0]  + b*P[1]   + c*P[2]   + d)   /   (a**2  + b**2  +  c**2)


Q = P - lam * np.array([a,b,b])




print("Closest point on the wall: ", Q)
```

**now lets plot it**

```python
x = np.linspace(-5,   10,   20)
y = np.linspace(-5 ,  10,   20)


X,  Y   = np.meshgrid(x,y)



#we need to solve for Z
Z = (-a*X  - b*Y - d)  /  c


```

```python
from mpl_toolkits.mplot3d import Axes3D


fig = plt.figure()
ax = fig.add_subplot(111,   projection='3d')

#plot plane
ax.plot_surface(X, Y, Z,  alpha=0.3)



#plot person point
ax.scatter(P[0],  P[1],   P[2],   s=60)



#plot closest point on the plane
ax.scatter(Q[0],   Q[1],   Q[2],   s=60)

#plot perpendicular line
ax.plot(  P[0],    Q[0],  [P[1],  Q[1]]   ,  [P[2], Q[2]]  )



ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')


ax.set_title('Shortest distance from Point to Wall')

ax.text(P[0], P[1], P[2], " Person", color='blue')
ax.text(Q[0], Q[1], Q[2], " Closest Point", color='red')

plt.show()




```
